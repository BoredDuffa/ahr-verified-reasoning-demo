# Notebook 17 — fixed kernel feasibility and learning

This notebook tests whether a batched fixed positive feature expansion can preserve useful raw ViT content interactions and improve complete JEPA training efficiency. Exact local 3×3 patch interactions are retained. It replaces dynamic ACA pivots and their dense fallback; approximation fidelity and image quality must be measured separately.

**Run All launches real detached work by default.** Training-only operator cases and full-step efficiency gates run first. Only GPUs observed free of other active project work enter each training phase; existing jobs are preserved. A single available GPU runs all pairs sequentially under the same hard phase clock. Passing gates trigger one four-pair equal-time learning screen. A promising screen triggers a frozen power-derived fresh-seed confirmation; failed or incomplete gates stop early and keep official test closed.

Inputs: frozen completed Notebook15 `vit-geometry-001` ViT reference and the existing Food101 project data. Outputs: portable report, numerical/timing/accuracy receipts and immutable phase clocks in `artifacts/notebook_17/<RUN_ID>/`; full state in `checkpoints/<RUN_ID>/`. Purpose: test a concrete efficient JEPA replacement hypothesis, without presuming better accuracy.

Expected runtime is conditional: up to 2 h for feasibility, another 3 h for the 256 px screen or 4 h at 512 px, and up to 12 h for confirmation. Maximum 18 h at 512 px; most failed candidates stop during feasibility. These are hard caps, not a promise of completion. Scientific status remains UNKNOWN on cutoff or invalid evidence.

Saved outputs identify their PROFILE and RUN_ID. Validation-profile output is a software presentation fixture. Full-profile launch output records detached execution; the committed report and study status hold the research results.

The basis follows established [positive orthogonal random-feature attention](https://arxiv.org/abs/2009.14794). Exact local replacement is our controlled experimental choice. GPU efficiency is judged against actual optimized attention, because lower arithmetic complexity alone does not guarantee speed, as explained in [FlashAttention](https://arxiv.org/abs/2205.14135). No new classical multipole expansion, universal manifold property or accuracy theorem is claimed.
