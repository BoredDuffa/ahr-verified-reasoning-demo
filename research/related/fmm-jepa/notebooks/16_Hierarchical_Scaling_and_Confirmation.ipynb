```python
from pathlib import Path
import json,os,sys,types
os.environ.setdefault('CUBLAS_WORKSPACE_CONFIG',':4096:8')
ROOT=Path('/home/benjamin/workspace/ahr_go/fmm_jeppa')

def load16_workflow(root=ROOT):
    path=Path(root)/'experiments/16_hierarchical_confirmation/workflow.ipynb'
    module=types.ModuleType('fmm16_frontdoor_workflow');module.__file__=str(path)
    for cell in json.loads(path.read_text())['cells']:
        if cell['cell_type']=='code' and 'ahr-export' in cell.get('metadata',{}).get('tags',[]):
            exec(compile(''.join(cell['source']),str(path)+'#ahr-export','exec'),module.__dict__)
    module.C.bootstrap(root)
    return module
```

The settings below select the immutable run identity. For a software-only fresh-kernel check set `FMM16_PROFILE=validation` before executing. The default is the actual full pipeline.

```python
RUN_ID=os.environ.get('FMM16_RUN_ID','hierarchical-confirmation-001')
PROFILE=os.environ.get('FMM16_PROFILE','full')
SOURCE_RUN=os.environ.get('FMM16_SOURCE_RUN') or None
HARD_CAP_SECONDS=12*60*60
WORKFLOW=load16_workflow()
print('Profile:',PROFILE,' | Run:',RUN_ID)
print('Run All launches measured learned-token checks, then only permitted confirmation stages.')
```

```python
LAUNCH=WORKFLOW.launch_study(ROOT,RUN_ID,profile=PROFILE,source_run=SOURCE_RUN,max_seconds=HARD_CAP_SECONDS)
print(json.dumps(LAUNCH,indent=2,default=str))
```
