```python
import os
import json
import shutil
import kagglehub

# 1) Load credentials from your local kaggle.json and set them
kaggle_json_path = "/home/benjamin/.1AHR_maths&words/kaggle.json"
with open(kaggle_json_path, "r") as f:
    kaggle_creds = json.load(f)

# kagglehub looks for these two environment variables by default
os.environ["KAGGLE_USERNAME"] = kaggle_creds["username"]
os.environ["KAGGLE_KEY"] = kaggle_creds["key"]

# 2) Set your exact destination folder
dest_path = "/home/benjamin/.1AHR_maths&words/data_aimo"

# 3) This downloads ALL competition data files into the kagglehub cache
print("Authentication successful! Starting download...")
cache_path = kagglehub.competition_download("ai-mathematical-olympiad-progress-prize-3")
print("Downloaded to cache:", cache_path)

# 4) Copy the files from the cache into your specific folder
os.makedirs(dest_path, exist_ok=True)
shutil.copytree(cache_path, dest_path, dirs_exist_ok=True)

print("Data saved to your folder:", dest_path)
```
